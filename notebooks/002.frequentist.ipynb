# package

```python
import pandas as pd
import statsmodels.formula.api as smf
```

# data

```python
PATH = './data/housing.csv'
df = pd.read_csv(PATH)
```

```python
df["YEARMONTH"] = pd.to_datetime(
    df["YEARMONTH"].astype(str),
    format="%Y-%m",
)
```

# frequentist DiD

```python
FUNCTION = 'PRICE ~ C(REGION_IDX) + C(TIME_IDX) + TREATMENT:PERIOD_1 + TREATMENT:PERIOD_2'

kwargs = dict(
    formula=FUNCTION,
    data=df,
)

model = smf.ols(**kwargs)
```

```python
result = model.fit()
```

```python
print(result.summary())
```

# frequentist DiD (clustered standard error)

```python
FUNCTION = 'PRICE ~ C(REGION_IDX) + C(TIME_IDX) + TREATMENT:PERIOD_1 + TREATMENT:PERIOD_2'

kwargs = dict(
    formula=FUNCTION,
    data=df,
)

model = smf.ols(**kwargs)
```

```python
kwargs = dict(
    cov_type='cluster', 
    cov_kwds={'groups': df['REGION_IDX']},
)

result = model.fit(**kwargs)
```

```python
print(result.summary())
```
